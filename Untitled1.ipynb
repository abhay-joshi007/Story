```python
!pip install textblob nltk seaborn

import pandas as pd
import re
import nltk
import seaborn as sns
import matplotlib.pyplot as plt

from textblob import TextBlob
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

nltk.download('stopwords')
```

```python
df = pd.read_csv("data_science.csv")

# Use only tweet column
df = df[['tweet']]
print(df.head())
```

```python
def get_sentiment(text):
    score = TextBlob(str(text)).sentiment.polarity
    if score > 0:
        return 1   # Positive
    else:
        return 0   # Negative

df['sentiment'] = df['tweet'].apply(get_sentiment)
```

```python
ps = PorterStemmer()
stop_words = set(stopwords.words('english'))

def clean_text(text):
    text = re.sub('[^a-zA-Z]', ' ', str(text))
    text = text.lower()
    words = text.split()
    words = [ps.stem(w) for w in words if w not in stop_words]
    return " ".join(words)

df['cleaned'] = df['tweet'].apply(clean_text)
```

```python
tfidf = TfidfVectorizer(max_features=5000)
X = tfidf.fit_transform(df['cleaned'])
y = df['sentiment']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
model = MultinomialNB()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

```python
cm = confusion_matrix(y_test, y_pred)

sns.heatmap(cm, annot=True, fmt='d')
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")
plt.show()
```

```python
def predict(text):
    cleaned = clean_text(text)
    vec = tfidf.transform([cleaned])
    pred = model.predict(vec)[0]
    return "Positive" if pred == 1 else "Negative"

print(predict("I love this project"))
print(predict("This is very bad"))
```

```python
# (for demo)

print("Example 1:")
print("Tweet: I absolutely love working with data science tools!")
print("Prediction:", predict("I absolutely love working with data science tools!"))
print()

print("Example 2:")
print("Tweet: This model is terrible and gives wrong predictions.")
print("Prediction:", predict("This model is terrible and gives wrong predictions."))
print()

print("Example 3:")
print("Tweet: The results are okay but could be better.")
print("Prediction:", predict("The results are okay but could be better."))
```
